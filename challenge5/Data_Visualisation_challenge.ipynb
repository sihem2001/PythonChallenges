# 2. Basic Understanding
## Feature description

### Describe the features in the dataset. What kind of information do they represent

### Dataset Feature Descriptions

- **id**: Unique identifier for each patient/sample.  
- **diagnosis**: Tumor classification — `M` = malignant, `B` = benign.

#### Mean (average) measurements of the tumor
- **radius_mean**: Mean distance from center to perimeter.  
- **texture_mean**: Average variation in gray-scale pixel intensity.  
- **perimeter_mean**: Mean perimeter length of the tumor boundary.  
- **area_mean**: Mean area of the tumor.  
- **smoothness_mean**: Average local variation in radius length (surface smoothness).  
- **compactness_mean**: Mean of (perimeter² / area - 1.0).  
- **concavity_mean**: Mean severity of concave portions of the contour.  
- **concave points_mean**: Mean number of concave boundary points.  
- **symmetry_mean**: Average symmetry of the tumor shape.  
- **fractal_dimension_mean**: Mean complexity of the boundary.

#### Standard error (variation) of the same features
- **radius_se**: Standard error of radius.  
- **texture_se**: Standard error of texture.  
- **perimeter_se**: Standard error of perimeter.  
- **area_se**: Standard error of area.  
- **smoothness_se**: Standard error of smoothness.  
- **compactness_se**: Standard error of compactness.  
- **concavity_se**: Standard error of concavity.  
- **concave points_se**: Standard error of concave points.  
- **symmetry_se**: Standard error of symmetry.  
- **fractal_dimension_se**: Standard error of fractal dimension.

#### “Worst” (largest) values observed in the tumor
- **radius_worst**: Largest observed radius.  
- **texture_worst**: Largest observed texture variation.  
- **perimeter_worst**: Largest observed perimeter value.  
- **area_worst**: Largest observed tumor area.  
- **smoothness_worst**: Worst (highest) smoothness value.  
- **compactness_worst**: Worst compactness value.  
- **concavity_worst**: Worst concavity value.  
- **concave points_worst**: Worst count of concave points.  
- **symmetry_worst**: Worst symmetry score.  
- **fractal_dimension_worst**: Highest fractal dimension observed.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_theme(style="ticks")

# Get all feature columns
feature_columns = df.columns.drop(['id', 'diagnosis_Result', 'Unnamed: 32']).tolist()

# Manual standardization (Z-score)
df_standardized = df.copy()
for feature in feature_columns:
    mean_val = df[feature].mean()
    std_val = df[feature].std()
    df_standardized[feature] = (df[feature] - mean_val) / std_val

# Create long-format dataframe
plot_data = pd.melt(df_standardized, 
                    id_vars=['diagnosis_Result'], 
                    value_vars=feature_columns,
                    var_name='feature', 
                    value_name='value')

# Create the plot
f, ax = plt.subplots(figsize=(20, 20))

sns.violinplot(
    data=plot_data, 
    x="value", 
    y="feature", 
    hue="diagnosis_Result",
    split=True,
    inner="quartile",
    palette="vlag",
    scale="width"
)

ax.xaxis.grid(True)
ax.set(ylabel="Features", xlabel="Z-score (Standardized Value)")
ax.axvline(x=0, color='gray', linestyle='--', alpha=0.5, label='Mean')
ax.set_xlim(-5, 5)  # Set reasonable limits for standardized data
sns.despine(trim=True, left=True)

plt.title("Standardized Feature Distributions (Z-scores) by Diagnosis Result")
plt.tight_layout()
plt.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
from sklearn.datasets import load_breast_cancer
import numpy as np

# Load breast cancer dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
df['diagnosis'] = df['target'].map({0: 'Malignant', 1: 'Benign'})

print("Dataset shape:", df.shape)
print("Features:", data.feature_names[:10])  # Show first 10 features

# Create 3D subplots
fig = make_subplots(
    rows=2, cols=2,
    specs=[[{'type': 'scatter3d'}, {'type': 'scatter3d'}],
           [{'type': 'scatter3d'}, {'type': 'scatter3d'}]],
    subplot_titles=(
        'Mean Features: Texture vs Area vs Smoothness',
        'Worst Features: Texture vs Area vs Smoothness',
        'Mean Features: Compactness vs Concavity vs Concave Points',
        'Worst Features: Compactness vs Concavity vs Concave Points'
    ),
    vertical_spacing=0.08,
    horizontal_spacing=0.08
)

# Color mapping for diagnoses
color_map = {'Malignant': 'red', 'Benign': 'blue'}

malignant = df[df['diagnosis'] == 'Malignant']
benign = df[df['diagnosis'] == 'Benign']

fig.add_trace(
    go.Scatter3d(
        x=malignant['mean texture'],
        y=malignant['mean area'],
        z=malignant['mean smoothness'],
        mode='markers',
        marker=dict(size=4, color=color_map['Malignant'], opacity=0.7),
        name='Malignant',
        legendgroup='Malignant',
        showlegend=True
    ),
    row=1, col=1
)

fig.add_trace(
    go.Scatter3d(
        x=benign['mean texture'],
        y=benign['mean area'],
        z=benign['mean smoothness'],
        mode='markers',
        marker=dict(size=4, color=color_map['Benign'], opacity=0.7),
        name='Benign',
        legendgroup='Benign',
        showlegend=True
    ),
    row=1, col=1
)

fig.add_trace(
    go.Scatter3d(
        x=malignant['worst texture'],
        y=malignant['worst area'],
        z=malignant['worst smoothness'],
        mode='markers',
        marker=dict(size=4, color=color_map['Malignant'], opacity=0.7),
        name='Malignant',
        legendgroup='Malignant',
        showlegend=False
    ),
    row=1, col=2
)

fig.add_trace(
    go.Scatter3d(
        x=benign['worst texture'],
        y=benign['worst area'],
        z=benign['worst smoothness'],
        mode='markers',
        marker=dict(size=4, color=color_map['Benign'], opacity=0.7),
        name='Benign',
        legendgroup='Benign',
        showlegend=False
    ),
    row=1, col=2
)

fig.add_trace(
    go.Scatter3d(
        x=malignant['mean compactness'],
        y=malignant['mean concavity'],
        z=malignant['mean concave points'],
        mode='markers',
        marker=dict(size=4, color=color_map['Malignant'], opacity=0.7),
        name='Malignant',
        legendgroup='Malignant',
        showlegend=False
    ),
    row=2, col=1
)

fig.add_trace(
    go.Scatter3d(
        x=benign['mean compactness'],
        y=benign['mean concavity'],
        z=benign['mean concave points'],
        mode='markers',
        marker=dict(size=4, color=color_map['Benign'], opacity=0.7),
        name='Benign',
        legendgroup='Benign',
        showlegend=False
    ),
    row=2, col=1
)
fig.add_trace(
    go.Scatter3d(
        x=malignant['worst compactness'],
        y=malignant['worst concavity'],
        z=malignant['worst concave points'],
        mode='markers',
        marker=dict(size=4, color=color_map['Malignant'], opacity=0.7),
        name='Malignant',
        legendgroup='Malignant',
        showlegend=False
    ),
    row=2, col=2
)

fig.add_trace(
    go.Scatter3d(
        x=benign['worst compactness'],
        y=benign['worst concavity'],
        z=benign['worst concave points'],
        mode='markers',
        marker=dict(size=4, color=color_map['Benign'], opacity=0.7),
        name='Benign',
        legendgroup='Benign',
        showlegend=False
    ),
    row=2, col=2
)

fig.update_layout(
    title_text="Breast Cancer Dataset - 3D Feature Visualizations",
    height=1000,
    legend=dict(
        yanchor="top",
        y=0.99,
        xanchor="left",
        x=0.01
    )
)

fig.update_scenes(
    xaxis_title_text="Mean Texture",
    yaxis_title_text="Mean Area", 
    zaxis_title_text="Mean Smoothness",
    row=1, col=1
)

fig.update_scenes(
    xaxis_title_text="Worst Texture",
    yaxis_title_text="Worst Area",
    zaxis_title_text="Worst Smoothness",
    row=1, col=2
)

fig.update_scenes(
    xaxis_title_text="Mean Compactness",
    yaxis_title_text="Mean Concavity",
    zaxis_title_text="Mean Concave Points",
    row=2, col=1
)

fig.update_scenes(
    xaxis_title_text="Worst Compactness",
    yaxis_title_text="Worst Concavity", 
    zaxis_title_text="Worst Concave Points",
    row=2, col=2
)

# Show the plot
fig.show()
```
