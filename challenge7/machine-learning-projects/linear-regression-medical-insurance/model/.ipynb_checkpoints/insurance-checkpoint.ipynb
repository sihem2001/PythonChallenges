```python
from sklearn.linear_model import LinearRegression, RidgeCV, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import pandas as pd

# Define models
alphas = [0.1, 1.0, 10.0, 50.0, 100.0]
models = {
    "Linear Regression": LinearRegression(),
    "RidgeCV": RidgeCV(alphas=alphas),
    "Lasso": Lasso(alpha=0.1),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42, n_estimators=100)
}

# Dictionary to store metrics
metrics = {"Model": [], "MSE": [], "RMSE": [], "R²": []}

# Train, predict, and evaluate each model
for name, model in models.items():
    model.fit(X_train_scaled, y_train)                # Train
    y_pred = model.predict(X_test_scaled)            # Predict
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    metrics["Model"].append(name)
    metrics["MSE"].append(mse)
    metrics["RMSE"].append(rmse)
    metrics["R²"].append(r2)

# Convert to DataFrame for easy viewing
metrics_df = pd.DataFrame(metrics)
print(metrics_df)
```

# Conclusion

### Linear Regression, Ridge, and Lasso

perform similarly, explaining about 79% of the variance, with moderate prediction errors (~5,581).
Regularization had little effect because the dataset is small and features are not highly correlated.

### Decision Tree

overfits slightly, showing higher error (RMSE ~6,131) and lower R² (~0.75).

### Random Forest

performs best, achieving the lowest RMSE (~4,775) and highest R² (~0.85), 
thanks to averaging multiple trees, which improves generalization and reduces overfitting.

## And he is the most accurate model for predicting insurance charges in this dataset.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

# List of models and predictions
model_preds = {
    "Linear Regression": LinearRegression().fit(X_train_scaled, y_train).predict(X_test_scaled),
    "RidgeCV": RidgeCV(alphas=[0.1, 1.0, 10.0, 50.0, 100.0]).fit(X_train_scaled, y_train).predict(X_test_scaled),
    "Lasso": Lasso(alpha=0.1).fit(X_train_scaled, y_train).predict(X_test_scaled),
    "Decision Tree": DecisionTreeRegressor(random_state=42).fit(X_train_scaled, y_train).predict(X_test_scaled),
    "Random Forest": RandomForestRegressor(random_state=42, n_estimators=100).fit(X_train_scaled, y_train).predict(X_test_scaled)
}

# Plot each model
for name, y_pred in model_preds.items():
    plt.scatter(y_test, y_pred, alpha=0.6, label=name)

# Plot perfect prediction line
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         'k--', lw=2, label="Perfect Prediction")

plt.xlabel("Actual Charges")
plt.ylabel("Predicted Charges")
plt.title("Predicted vs Actual Insurance Charges")
plt.legend()
plt.show()
```

# Cross-validation

```python
from sklearn.model_selection import cross_val_score
# 5-fold CV for Random Forest
rf_scores = cross_val_score(forest_reg, X_train_scaled, y_train, 
                            scoring="r2", cv=5)

print("Random Forest CV R² scores:", rf_scores)
print("Mean CV R²:", rf_scores.mean())
```

# Fine-tuning models (Hyperparameter tuning)

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [100, 200, 500],
    "max_depth": [None, 5, 10, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4]
}

grid_search = GridSearchCV(RandomForestRegressor(random_state=42),
                           param_grid, 
                           cv=5, 
                           scoring="r2",
                           n_jobs=-1)

grid_search.fit(X_train_scaled, y_train)

print("Best params:", grid_search.best_params_)
print("Best CV R²:", grid_search.best_score_)
```

<h3>Preprocessing & Handling Categorical Features:</h3>

<h3>Model Training & Evaluation:</h3>
<ul>
  <li>Trained multiple models: Linear Regression, Ridge, Lasso, Decision Tree, Random Forest.</li>
  <li>Evaluated using <strong>MSE, RMSE, and R²</strong>.</li>
  <li>Observed that Linear Regression, Ridge, and Lasso performed similarly (~R²=0.79).</li>

<h3>Hyperparameter Tuning:</h3>

</ul>

<h3>Final Insights & Interpretation:</h3>
<ul>
  <li>Random Forest is the most accurate model for this dataset.</li>
