```python
# Incorporate data
df = pd.read_csv('C:/Users/admin/firstcours/challengesPython/challenge5/data.csv')
df
```

```python
app = Dash()
```

```python
app.layout = [
    html.Div(children = 'My first App with Data',style={'textAlign': 'center', 'color': '#FF3F7F', 'fontSize': 30}),
    dash_table.DataTable(data=df.to_dict('records'),page_size=100 ),
    # some graphics
    html.Div(children = 'Visualizing Data -- histogram -- ', style={'textAlign': 'center', 'color': '#FF3F7F', 'fontSize': 30}),
    dcc.Graph(figure=px.histogram(df, x='diagnosis', y='radius_mean', histfunc='avg')),
    
    html.Div(children =  ' Controls and Callbacks',style={'textAlign': 'center', 'color': '#FF3F7F', 'fontSize': 30}),
    html.Hr(),
    dcc.RadioItems(options=[
        {'label': 'Radius Mean', 'value':'radius_mean'},
         {'label': 'Texture Mean', 'value': 'texture_mean'},
         {'label': 'Area Mean', 'value': 'area_mean'}
    ], 
      value='radius_mean', 
        id='feature-selector'
    ),
   dcc.Graph(figure={}, id='cancer-graph')
]
@callback(
    Output(component_id='cancer-graph', component_property='figure'),
     Input(component_id='feature-selector', component_property='value')
)
def update_graph(selected_feature):
    fig = px.histogram(df, x='diagnosis', y=selected_feature, histfunc='avg')
    return fig
    

    
```

```python
if __name__ == '__main__':
    app.run(debug=True)
```
