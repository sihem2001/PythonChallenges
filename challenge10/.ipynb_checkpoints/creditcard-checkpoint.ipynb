# Credit Card Fraud Detection - Comparative study for XGboost and AdaBoost

## Dataset Selection
It's a dataset for classification problem to classify if a transaction is fraud or normale based on the history of transaction.
V1 to V28 are the result of application PCA on variables (montant, type de paiement, heure .... ) they become component.
#### PCA : 
principal component analysis : this allowed to have same forme for all variables and protect data and make it simple for ml model.
#### ------------------------------------------------------------------------------------------------------------------------------
#### Caracteristique of dataset : 
284808 rows with columns 31 columns
Traget columns "Class": 1 for fraudulent transactions, 0 otherwise

#### Outliers

```python
# check if we have outliers 
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=data['Amount'])
plt.title("Outliers in Amount")
plt.show()
```

#### Visualisation 
Distribution des données  , corrélations avec Target, Boxplots pour outillers , analyse le déséquilibre des classes  

```python
# class imbalance (fraud is very rare)
import seaborn as sns 
import matplotlib.pyplot as plt

sns.countplot(x='Class',data=data, color = 'blue')
plt.title("Distribution of classes (0 = no fraud , 1 = fraud)")
plt.show()

print(data['Class'].value_counts(normalize=True) * 100)
```

```python
# distribution of V along Amount
plt.figure(figsize=(10,5))
sns.histplot(data['Amount'], bins=50)
plt.title("Transaction Amount Distribution")
plt.show()
```

```python
#correlation 
plt.figure(figsize=(28,28))
corr = data.corr()
sns.heatmap(corr , cmap='coolwarm', center=0 , annot=True,)
plt.title("Correlation with all features")
plt.savefig("heatmapCard.png")
plt.show()
```

```python
corr = data.corr()

plt.figure(figsize=(8,10))
sns.heatmap(corr[['Class']].sort_values(by='Class', ascending=False),
            annot=True, cmap='coolwarm')

plt.title("Correlation with Class (Fraud)")
plt.savefig("heatmapClass.png")
plt.show()
```

as we can see the most correlated feature are V11 V4 V2 but slightly not that strong to predict the fraud. This is what we need from xgboost and adabost to work on , with little feature that can lead as to predict we get a result
